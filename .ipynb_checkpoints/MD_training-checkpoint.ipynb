```python
import MDAnalysis
from MDAnalysis.tests.datafiles import PSF, DCD # test t trajectory
import numpy.linalg
# loading trajectory and topology into a universe U
u = MDAnalysis.Universe(PSF,DCD)
# from 4AKE segid(segment ID), select first atome named N and last atom named C
nterm = u.select_atoms('segid 4AKE and name N')[0]
cterm = u.select_atoms('segid 4AKE and name C')[-1]
# select backbone atoms ( atom groups)
bb = u.select_atoms('protein and backbone')
# iterating through all the frames
for ts in u.trajectory:
    r = cterm.position - nterm.position # vector r connecting the positions of the first atom and the last atom
    d = numpy.linalg.norm(r) # calculating the distance b/n both the atoms
    rgyr = bb.radius_of_gyration() # method of AtomGroup
    print("frame ={0}: d ={1} A, Rgyr = {2} A".format(ts.frame, d, rgyr))
```

atom_resindex (list): list of resindices for each atom

residue_segindex (list): list of segindices for each residue

trajectory (bool): whether to attach a MemoryReader trajectory (default False)

velocities (bool): whether to include velocities in the trajectory (default False)
