```python
import MDAnalysis as mda
import numpy as np
```

```python
'''
Loding files into u
'''
u = mda.Universe("test.pdb")
```

```python
'''
Defining the selection of atoms based on the system
'''
dye1 = u.select_atoms("resname DYE1")
dye2 = u.select_atoms("resname DYE2")

# for angle: we pick te=wo atoms per dye and define them as a vector
dye1_atoms = u.select_atoms("resname DYE1 and atoms A1 A2")
dye2_atoms = u.select_atoms("resname DYE2 and atoms B1 B2")
```

```python
times = []
distances = []
angles = []

for ts in u.trajectory:
    times.append(ts.time) # time is appended into the time matrix
    cog1 = dye1.center_of_geometry() # COG of dye 1
    cog2 = dye2.center_of_geometry() # COG of dye 2
    dist = np.linalg.norm(cog1-cog2) # distance b/n dye 1 and dye 2
    distances.append(dist) # appending the distances matrix
    #vectors b/n both the dye positions
    v1 = dye1_atoms.positions[1] - dye1_atoms.positions[0]
    v2 = dye2_atoms.positions[1] - dye2_atoms.positions[0]
    #normalizing the vectors gives
    v1 = v1 / np.linalg.norm(v1)
    v2 = v2 / np.linalg.norm(v2)
    # angle is cos theta
    cos_theta = np.dot(v1, v2)
    angle = np.degrees(np.arccos(cos_theta))

    angles.append(angle)
    # printing the outputs
    print(f"{time:.3f} {dist:.3f} {angle:.2f}")
print("DONE.")    
```
